```python

# SISTEMA DE GESTÃO DE NOTAS DE ALUNOS
# Linguagem de Programação - Python

# Função responsável por cadastrar as notas
def cadastrar_notas():
    notas = []

    print(" CADASTRO DE NOTAS ")

    # Solicita a quantidade de notas que serão cadastradas
    while True:
        quantidade = float(input("Quantas notas deseja cadastrar? "))

        # Verifica se a quantidade é um número inteiro e maior que zero
        if quantidade.is_integer() and quantidade > 0:
            quantidade = int(quantidade)
            break
    else:
            print("Número inválido! Digite uma quantidade inteira de notas.")

    # Repete a solicitação de nota conforme a quantidade informada
    for i in range(quantidade):

        while True:
            nota = float(input(f"Digite a {i + 1}ª nota (0 a 10): "))

            # Verifica se a nota está dentro do intervalo permitido
            if 0 <= nota <= 10:
                notas.append(nota)
                break
            else:
                print("Nota inválida! Digite uma nota entre 0 e 10.")

    return notas


# Função responsável por calcular a média
def calcular_media(notas):
    media = sum(notas) / len(notas)
    return media


# Função responsável por determinar a situação do aluno
def determinar_situacao(media):

    # Se a média for maior ou igual a 6, o aluno está aprovado
    if media >= 6:
        return "Aprovado"

    # Caso contrário, o aluno está reprovado
    else:
        return "Reprovado"


# Função responsável por exibir o relatório final
def exibir_relatorio(notas, media, situacao):

    print("\n================================")
    print("        RELATÓRIO FINAL")
    print("================================")

    print("Notas inseridas:", notas)
    print(f"Média do aluno: {media:.2f}")
    print("Situação:", situacao)

    print("================================")


# ============================================
# PROGRAMA PRINCIPAL
# ============================================

# Chama a função para cadastrar as notas
notas = cadastrar_notas()

# Calcula a média das notas cadastradas
media = calcular_media(notas)

# Determina se o aluno foi aprovado ou reprovado
situacao = determinar_situacao(media)

# Exibe o relatório final
exibir_relatorio(notas, media, situacao)
```
