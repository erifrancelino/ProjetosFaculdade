```python

# Machine Learning aplicado à loja ALUCINA

import tensorflow as tf
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix

# Base fictícia de produtos artesanais da ALUCINA
dados = {
    "produto": [
        "Mini xicara", "Vaso artesanal", "Peca autoral",
        "Mini prato", "Chaveiro cogumelo", "Chaveiro xicara",
        "Chaveiro artesanal", "Pingente artesanal",
        "Marcador de pagina", "Baralho autoral",
        "Cartas ilustradas", "Kit de jogos",
        "Mini escultura", "Chaveiro personalizado",
        "Marcador ilustrado"
    ],
    "preco": [
        25, 70, 120, 40, 15, 18, 20, 22,
        10, 60, 35, 80, 90, 25, 12
    ],
    "peso_g": [
        80, 350, 500, 150, 15, 20, 25, 18,
        10, 180, 100, 250, 400, 30, 12
    ],
    "tamanho_cm": [
        5, 12, 18, 8, 4, 4, 5, 3,
        15, 10, 9, 15, 14, 5, 16
    ],
    "categoria": [
        "Ceramica", "Ceramica", "Ceramica", "Ceramica",
        "Acessorios", "Acessorios", "Acessorios",
        "Acessorios", "Papelaria_Jogos", "Papelaria_Jogos",
        "Papelaria_Jogos", "Papelaria_Jogos",
        "Ceramica", "Acessorios", "Papelaria_Jogos"
    ]
}

df_alucina = pd.DataFrame(dados)

# Visualizar a base de dados
display(df_alucina)

print("Total de produtos:", len(df_alucina))
print("Categorias:", df_alucina["categoria"].unique())

# Selecionar as características numéricas utilizadas pelo modelo
X = df_alucina[["preco", "peso_g", "tamanho_cm"]].values

# Transformar as categorias em números
categorias = {
    "Acessorios": 0,
    "Ceramica": 1,
    "Papelaria_Jogos": 2
}

y = df_alucina["categoria"].map(categorias).values

# Dividir em treinamento e teste, preservando as categorias
X_treino, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.33,
    random_state=42,
    stratify=y
)

# Normalizar as características
normalizador = StandardScaler()
X_treino = normalizador.fit_transform(X_treino)
X_teste = normalizador.transform(X_teste)

print("Produtos para treinamento:", len(X_treino))
print("Produtos para teste:", len(X_teste))
print("Preparação dos dados concluída!")


# Criar o modelo de classificação
modelo = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(3,)),
    tf.keras.layers.Dense(12, activation="relu"),
    tf.keras.layers.Dense(8, activation="relu"),
    tf.keras.layers.Dense(3, activation="softmax")
])

# Configurar o treinamento
modelo.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

modelo.summary()

# Treinar a rede neural
historico = modelo.fit(
    X_treino,
    y_treino,
    epochs=100,
    batch_size=4,
    verbose=1
)

print("Treinamento concluído!")

# Avaliar o desempenho nos dados de teste
perda, acuracia = modelo.evaluate(
    X_teste,
    y_teste,
    verbose=0
)

print(f"Acurácia no teste: {acuracia * 100:.2f}%")
print(f"Perda no teste: {perda:.4f}")

# Gerar previsões para os produtos de teste
probabilidades = modelo.predict(X_teste, verbose=0)
y_previsto = np.argmax(probabilidades, axis=1)

nomes_categorias = [
    "Acessorios",
    "Ceramica",
    "Papelaria_Jogos"
]

print("\nRelatório de classificação:")
print(classification_report(
    y_teste,
    y_previsto,
    labels=[0, 1, 2],
    target_names=nomes_categorias,
    zero_division=0
))

print("Matriz de confusão:")
print(confusion_matrix(
    y_teste,
    y_previsto,
    labels=[0, 1, 2]
))

# Exemplo de um novo produto:
# preço de R$ 18, peso de 20 g e tamanho de 4 cm
novo_produto = np.array([[18, 20, 4]])

# Aplicar a mesma normalização usada no treinamento
novo_produto_normalizado = normalizador.transform(novo_produto)

# Prever a categoria
probabilidades = modelo.predict(
    novo_produto_normalizado,
    verbose=0
)[0]

indice = np.argmax(probabilidades)
categoria_prevista = nomes_categorias[indice]

print("Novo produto: chaveiro artesanal")
print("Categoria prevista:", categoria_prevista)
print("\nProbabilidades estimadas:")

for nome, probabilidade in zip(nomes_categorias, probabilidades):
    print(f"{nome}: {probabilidade * 100:.2f}%")

```
