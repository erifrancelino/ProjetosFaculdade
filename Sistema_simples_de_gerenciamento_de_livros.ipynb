```python

# ============================================================
# ROTEIRO DE AULA PRÁTICA
# Disciplina: Linguagem de Programação
# Unidade 2: Explorando Recursos do Python
# Aula 4: Bibliotecas e Módulos em Python
# Sistema simples de gerenciamento de livros
# ============================================================

# Importação da biblioteca Matplotlib para gerar gráficos
import matplotlib.pyplot as plt


# PASSO 1: Definir a classe Livro
class Livro:
    # O método construtor inicializa os atributos de cada livro
    def __init__(self, titulo, autor, genero, quantidade):
        self.titulo = titulo
        self.autor = autor
        self.genero = genero
        self.quantidade = quantidade

    # Método para apresentar os dados do livro
    def exibir_dados(self):
        return (
            f"Título: {self.titulo} | "
            f"Autor: {self.autor} | "
            f"Gênero: {self.genero} | "
            f"Quantidade disponível: {self.quantidade}"
        )


# PASSO 2: Criar uma lista vazia para armazenar os livros
livros = []


# PASSO 3.1: Função para cadastrar um novo livro
def cadastrar_livro(titulo, autor, genero, quantidade):
    # Verifica se a quantidade informada é válida
    if not isinstance(quantidade, int) or quantidade < 0:
        print("Erro: a quantidade deve ser um número inteiro não negativo.")
        return

    # Cria um objeto da classe Livro
    novo_livro = Livro(titulo, autor, genero, quantidade)

    # Adiciona o objeto à lista de livros
    livros.append(novo_livro)

    print(f'Livro "{titulo}" cadastrado com sucesso!')


# PASSO 3.2: Função para listar todos os livros
def listar_livros():
    if not livros:
        print("Nenhum livro cadastrado.")
        return

    print("\n========== LIVROS CADASTRADOS ==========")

    for livro in livros:
        print(livro.exibir_dados())


# PASSO 3.3: Função para buscar um livro pelo título
def buscar_livro(titulo):
    # A busca ignora diferenças entre letras maiúsculas e minúsculas
    for livro in livros:
        if livro.titulo.casefold() == titulo.casefold():
            print("\nLivro encontrado:")
            print(livro.exibir_dados())
            return livro

    print(f'Livro "{titulo}" não encontrado.')
    return None


# PASSO 4: Gerar um gráfico da quantidade de livros por gênero
def gerar_grafico():
    # Dicionário que armazenará a quantidade total por gênero
    quantidades_por_genero = {}

    # Soma as quantidades dos livros de cada gênero
    for livro in livros:
        genero = livro.genero
        quantidades_por_genero[genero] = (
            quantidades_por_genero.get(genero, 0) + livro.quantidade
        )

    # Verifica se existem livros cadastrados
    if not quantidades_por_genero:
        print("Cadastre livros antes de gerar o gráfico.")
        return

    # Separa os gêneros e suas quantidades
    generos = list(quantidades_por_genero.keys())
    quantidades = list(quantidades_por_genero.values())

    # Cria o gráfico de barras com Matplotlib
    plt.figure(figsize=(9, 5))
    barras = plt.bar(generos, quantidades)

    # Adiciona título e identificação dos eixos
    plt.title("Quantidade de Livros Disponíveis por Gênero")
    plt.xlabel("Gênero")
    plt.ylabel("Quantidade disponível")
    plt.bar_label(barras, padding=3)
    plt.tight_layout()

    # Exibe o gráfico
    plt.show()


# ============================================================
# PASSO 5: Testar o sistema com diferentes entradas
# ============================================================

# Cadastro de livros para testar o funcionamento
cadastrar_livro("Dom Casmurro", "Machado de Assis", "Romance", 5)
cadastrar_livro("O Hobbit", "J. R. R. Tolkien", "Fantasia", 3)
cadastrar_livro("1984", "George Orwell", "Ficção", 4)
cadastrar_livro("Orgulho e Preconceito", "Jane Austen", "Romance", 2)
cadastrar_livro("Percy Jackson", "Rick Riordan", "Fantasia", 6)

# Teste da listagem de todos os livros
listar_livros()

# Teste da busca por um título existente
buscar_livro("Dom Casmurro")

# Teste da busca por um título inexistente
buscar_livro("Livro inexistente")

# Teste da geração do gráfico por gênero
gerar_grafico()
```
